# Function Overloading

```python
def fun(var1, var2 = None):
    if var2 is not None:
        print(var1, var2)
    
    else:
        print(var1)
fun(10)
fun(10, 20)
```

```python
## example 2 : default arguments 

def calculator(a = 0, b = 0, c = 0, d = 0):
    return a + b + c + d
print(calculator(100, 200))
print(calculator(100, 200, 300,))
print(calculator(100, 200, 300, 400))
```

```python
## *args for function overloading

def calculator(*arg):   #*arg used for acessing multiple values
    return sum(arg)
    
print(calculator(100, 200))
print(calculator(100, 200, 300,))
print(calculator(100, 200, 300, 400))
```

```python
# function overloading using decorator

from multipledispatch import dispatch

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
   print(var1, var2)

fun(10)
fun(10, 20)
```

```python
@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
   print(var1, var2)

@dispatch(float, float, int)
def fun(f1, f2, f3):
    print(f1, f2, f3)

fun(10)
fun(10, 20)
fun(2.5, 4.1, 100)
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.Varity = variety
        self.Price = price
        
    def __add__(self, d2_val):
        total_price = self.Price + d2_val.Price
        return total_price
        
    def __sub__(self, d2_val):
       total_price = self.Price  - d2_val.Price
       return total_price

    def __mul__(self, d2_val):
       total_price = self.Price  * d2_val.Price
       return total_price

    def __truediv__(self, d2_val):
       total_price = self.Price  / d2_val.Price
       return total_price
       
    def __floordiv__(self, d2_val):
       total_price = self.Price  // d2_val.Price
       return total_price
       
    def __mod__(self, d2_val):
       total_price = self.Price  % d2_val.Price
       return total_price

       
d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1 + d2)
print(d1 - d2)
print(d1 * d2)
print(d2 / d1)  # truediv
print(d2 // d1) # floordiv
print(d1 % d2)  # mod
print(d1 * d2) # pow
```

```python
## comparision operator

class Dosa:
    def __init__(self, variety, price):
        self.Varity = variety
        self.Price = price
        
    def __eq__(self, d2_val):
        total_price = self.Price == d2_val.Price
        return total_price
        
    def __gt__(self, d2_val):
       total_price = self.Price  > d2_val.Price
       return total_price

    def __lt__(self, d2_val):
       total_price = self.Price  < d2_val.Price
       return total_price

    def __ge__(self, d2_val):
       total_price = self.Price  >= d2_val.Price
       return total_price
       
    def __le__(self, d2_val):
       total_price = self.Price <= d2_val.Price
       return total_price
       
    

          
        
d1 = Dosa("Masala", 60)
d2 = Dosa("IceCream", 110)

print(d1 > d2)
print(d1 < d2)
print(d1 == d2)
print(d1 >= d2)
print(d1 <= d2)
```

# Data Abstraction

```python
# Example-1 : how to define abstract class, abstract method 

from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): # abstract method
        pass

    @abstractmethod
    def LED_screen(Self):
        pass

    def insurance(self): # concrete method
        print("Please take Mahindra company insurance")

class hyd_mahindra(Head_Mahindra): # child class
    def air_bag(self): # abstract method
        print("I will use Nylon to design the Air bag")

    def LED_screen(Self):
        print("Sony LED")

class noida_mahindra(Head_Mahindra):
    def air_bag(self):
        print("I will use Cotton")

    def LED_screen(Self):
        print("Sony LED")


    

obj_hyd = hyd_mahindra() # create object of hyd child class
obj_noida = noida_mahindra

obj_hyd.LED_screen()
obj_hyd.insurance()
#obj_head = Head_Mahindra()
```
