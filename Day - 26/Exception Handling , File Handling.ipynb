# Exception Handling


```python
# Example: structure of try-except

try:
    print(x)
except:
    print("Some Error occur")

print("Hello")
a,b = 10, 20
print(a+b)
```

```python
# Example-2: structure of try-except

try:
    print(y)
except Exception as e:
    print(f"This is actual error : {e}")
    print(type(e))

print("Hello")
a,b = 10, 20
print(a+b)
```

```python
# Example-3: else block

try:
    y = 20
    print(y)
except Exception as e:
    print(f"compiler error : {e}")
    print(type(e))
else:
    print("Hurry, we successfully executed code")

print("Hello")
a,b = 10, 20
print(a+b)
```

```python
# Example-3: finally block: execute if NO error + Error

try:
    y = 20
    print(z)
except Exception as e:
    print(f"compiler error : {e}")
    print(type(e))
else:
    print("Hurry, we successfully executed code")

finally:
    print("I will execute whatever happen")
```

# type of errors

```python
# NameError : variable not defined
# ZeroDivisionError
try:
    print(z)
except NameError:
    print("Some Error occur")
except ZeroDivisionError:
    print("Some Error occur")
```

```python
try:
    y = 10
    print(y/0)
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
```

```python
try:
    l1 = [10, 20, 30]
    print(l1[5])
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
except IndexError:
    print("This is Index error")
except Exception as e:
    print(type(e),e)
```

```python
try:
    l1 = [10, 20, 30]
    print(l1[5])
except NameError:
    print("This is Name error")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
except IndexError:
    print("This is Index error")
except KeyError:
    print("This is KeyError")
except Exception as e:
    print(type(e),e)
```

# Use of Raise Keyword

```python
# raise is used to generate user defined error

age = -22

if age < 0:
    raise Exception("Sorry, Age cannot be in negative")
```

# File Handling


```python
import os

os.getcwd()
```

```python
# open("file_name", "mode").
# f is file object through which we can access the content of file
f = open("ds20_file.txt", "r")

for line in f:
    print(line)

f.close() # closing the file, deleting from RAM
```

```python
# with statement: this will automatically close 
with open("ds20_file.txt", "r") as f:
    for line in f:
        print(line)
```

```python
# read() : to get all lines from the file
with open("ds20_file.txt", "r") as f:
    print(f.read()) # read the entire file
```

```python
# readlines : to get all the lines into a list, to access particular line data
with open("ds20_file.txt", "r") as f:
    lines = f.readlines()
    print(lines)# read the entire file
```

```python
# accessing particular line number menstion num
with open("ds20_file.txt", "r") as f:
    lines = f.readlines(1)
    print(lines)# read the entire file
```

```python
with open("ds20_file.txt", "r") as f:
    lines = f.readlines(1)
    print(*lines) # unpacking of complete list
```

# writelines: writing multiple lines together in the file


```python
with open("pawan_kalyan123.txt", "a") as f:
    f.writelines(["dinesh have 100 gf . \n", "Still he is alive"])
```

```python
with open("pawan_kalyan123.txt", "r") as f:
    print(f.read())
```

```python
# w+ mode: write/read mode
# w+ mode will clear the pervious content
# w+ create new file if file don't exist

with open("pavan_reddy.txt", "w+") as f:
    f.write("pawan reddy file")
    f.seek(10)
    print(f.read())
```

```python
# r+ : read/write mode
# file must exist, otherwise shows error

with open("pavan_reddy.txt", "r+") as f:
    f.write("pawan reddy file and hardworking")
    f.seek(0)
    print(f.read())
```

```python
# a+ : Append/Read

with open("pavan_reddy.txt", "a+") as f:
    f.write("pawan reddy file")
    f.seek(0)
    print(f.read())
```
